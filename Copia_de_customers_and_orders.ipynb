**Question 1:** How many customers are in the file?
(as help, we have added some comments and starter code to help you structure the solution)

```python
from os.path import exists
import csv

# check if file can be found and open the file
datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit(
        "You should run the first code cell and download the dataset files!"
    )

# Count data rows: skip the header explicitly, then count the rest
with open(datafile, "r", encoding="utf-8") as fl:
    csvreader = csv.reader(fl, delimiter=",")
    next(csvreader, None)                       # descarta la cabecera de forma explícita
    num_customers = sum(1 for _ in csvreader)   # cuenta solo filas de datos

print(f"Customers in the file: {num_customers}")
print("All done!")
```

**Question 2:** In how many different states do the customers live in?

```python
# open file and init vars
# evaluate the contents of the file
from os.path import exists
import csv

datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit(
        "You should run the first code cell and download the dataset files!"
    )

with open(datafile, "r", encoding="utf-8") as fl:
    csvreader = csv.reader(fl, delimiter=",")
    header = next(csvreader)  # descarta la cabecera de forma explícita
    # localiza la columna 'State' por nombre, sin importar mayúsculas/espacios
    state_col = [h.strip().upper() for h in header].index("STATE")
    states = {
        row[state_col].strip().upper()   # normaliza
        for row in csvreader
        if row[state_col].strip()        # ignora valores vacíos
    }

print(f"States found: {sorted(states)}")   # verificación (opcional)
print(f"Number of different states: {len(states)}")  # <- respuesta a entregar
print("All done!")
```

**Question 3** What is the state with most customers?

```python
from os.path import exists
import csv
from collections import Counter

datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit(
        "You should run the first code cell and download the dataset files!"
    )

with open(datafile, "r", encoding="utf-8") as fl:
    customers_by_state = Counter(
        customer["State"].strip().upper()      # normaliza
        for customer in csv.DictReader(fl)
        if customer["State"].strip()           # ignora estados vacíos
    )

# mayor conteo; a igualdad, orden alfabético (desempate determinista)
top_state, top_count = min(
    customers_by_state.items(), key=lambda kv: (-kv[1], kv[0])
)

print(f"The state with most customers is {top_state} ({top_count} customers)")
print(f"All counts: {dict(customers_by_state)}")  # verificación
print("All done!")
```

**Question 4** What is the state with the least customers?

```python
# ============================================================
# QUESTION 4: What is the state with the least customers?
# ============================================================
from os.path import exists
import csv
from collections import Counter

# verificar que el archivo exista y abrirlo
archivo = "./sample_data/customers.csv"
if not exists(archivo):
    raise SystemExit(
        "Debes ejecutar la primera celda de código y descargar los archivos del dataset!"
    )

with open(archivo, "r", encoding="utf-8") as fl:
    lector_csv = csv.reader(fl, delimiter=",")
    cabecera = next(lector_csv)                            # salta la fila de cabecera
    col_estado = [h.strip().upper() for h in cabecera].index("STATE")  # "STATE" es el nombre real de la columna en el archivo
    clientes_por_estado = Counter(
        fila[col_estado].strip().upper()                   # normaliza
        for fila in lector_csv
        if fila[col_estado].strip()                        # ignora estados vacíos
    )

minimo = min(clientes_por_estado.values())
estados_con_menos = sorted(
    e for e, n in clientes_por_estado.items() if n == minimo  # todos los empatados
)
print(f"Los estados con menos clientes son: {', '.join(estados_con_menos)} ({minimo} cliente cada uno)")
print("¡Listo!")
```

**Question 5:** What is the most common last name?

```python
from os.path import exists
import csv
from collections import Counter

# verificar que el archivo exista y abrirlo
archivo = "./sample_data/customers.csv"
if not exists(archivo):
    raise SystemExit(
        "Debes ejecutar la primera celda de código y descargar los archivos del dataset!"
    )

with open(archivo, "r", encoding="utf-8") as fl:
    lector_csv = csv.DictReader(fl)
    clientes_por_apellido = Counter(
        cliente["LastName"].strip().title()    # normaliza
        for cliente in lector_csv
        if cliente["LastName"].strip()         # ignora apellidos vacíos
    )

maximo = max(clientes_por_apellido.values())
apellidos_mas_comunes = sorted(
    a for a, n in clientes_por_apellido.items() if n == maximo  # todos los empatados
)
print(f"El apellido más común es: {', '.join(apellidos_mas_comunes)} ({maximo} clientes)")
print("¡Done!")
```

```python
# open and evaluate orders.csv
from os.path import exists
import csv
from collections import Counter, defaultdict

meses = {
    '01': 'January', '02': 'February', '03': 'March',    '04': 'April',
    '05': 'May',     '06': 'June',     '07': 'July',      '08': 'August',
    '09': 'September', '10': 'October', '11': 'November', '12': 'December'
}  # valores en inglés: son los datos que pide el quiz

# ---------- nombres de clientes: CustomerID -> "First Last" ----------
nombres = {}
ruta_clientes = "./sample_data/customers.csv"          # ruta original del curso
if exists(ruta_clientes):
    with open(ruta_clientes, "r", encoding="utf-8") as fl:
        lector = csv.DictReader(fl)     # lee por NOMBRE de columna, no por índice
        for cliente in lector:
            cid = (cliente.get("CustomerID") or "").strip()
            if cid and cid not in nombres:
                nombre = (
                    (cliente.get("FirstName") or "").strip().title()
                    + " "
                    + (cliente.get("LastName") or "").strip().title()
                ).strip()
                nombres[cid] = nombre

def etiqueta(cid):
    nombre = nombres.get(cid)
    return f"{cid} - {nombre}" if nombre else cid

# ---------- PASADA 1: agrupar filas por orden ----------
archivo_ordenes = "./sample_data/orders.csv"           # ruta original del curso
if not exists(archivo_ordenes):
    raise SystemExit(
        "Debes ejecutar la primera celda de código y descargar los archivos del dataset!"
    )

ordenes = {}  # OrderID -> {'cliente', 'fecha', 'total', 'items'}

with open(archivo_ordenes, "r", encoding="utf-8") as fl:
    for fila in csv.DictReader(fl):
        order_id = (fila["OrderID"] or "").strip()
        if not order_id:
            continue
        if order_id not in ordenes:                   # primera fila de la orden
            try:
                total = float((fila["OrderTotal"] or "").strip())
            except ValueError:
                total = None                          # OrderTotal sucio
            ordenes[order_id] = {
                "cliente": (fila["CustomerID"] or "").strip(),
                "fecha": (fila["Date"] or "").strip(),
                "total": total,
                "items": [(fila["ProductName"] or "").strip().lower()],
            }
        else:
            ordenes[order_id]["items"].append(
                (fila["ProductName"] or "").strip().lower()
            )

# ---------- PASADA 2: estadísticas dependientes de la fecha ----------
ordenes_por_mes = defaultdict(set)                    # (año, mes) -> {OrderID}
gasto_por_cliente = defaultdict(lambda: defaultdict(float))  # año -> cliente -> total
ventas_por_mes = defaultdict(float)                   # (año, mes) -> total vendido

for order_id, orden in ordenes.items():
    if not orden["fecha"]:
        continue                                      # sin fecha: solo cuenta en Q1-Q3
    try:
        anio, mes, _ = orden["fecha"].split(" ")[0].split("-")
    except ValueError:
        continue                                      # fecha malformada
    ordenes_por_mes[(anio, mes)].add(order_id)
    if orden["total"] is not None:
        gasto_por_cliente[anio][orden["cliente"]] += orden["total"]
        ventas_por_mes[(anio, mes)] += orden["total"]

# ---------- Respuestas ----------
items_por_orden = [len(o["items"]) for o in ordenes.values()]
q1 = len(ordenes)
q2 = sum(items_por_orden) / len(ordenes)
q3 = max(items_por_orden)
q4 = len(ordenes_por_mes.get(("2021", "10"), ()))

gasto_2021 = gasto_por_cliente.get("2021", {})
maximo_gasto = max(gasto_2021.values()) if gasto_2021 else 0.0
clientes_top = sorted(c for c, g in gasto_2021.items() if g == maximo_gasto)

ganadores_por_anio = {}
for anio in {a for a, _ in ventas_por_mes}:
    totales = {m: t for (a, m), t in ventas_por_mes.items() if a == anio}
    maximo = max(totales.values())
    ganadores_por_anio[anio] = sorted(m for m, t in totales.items() if t == maximo)[0]
frecuencia = Counter(ganadores_por_anio.values())
max_frec = max(frecuencia.values())
meses_mas_frecuentes = sorted(m for m, f in frecuencia.items() if f == max_frec)

ventas_historicas = defaultdict(float)
for (_, mes), total in ventas_por_mes.items():
    ventas_historicas[mes] += total
num_anios = len({a for a, _ in ordenes_por_mes})
promedio_mensual = {m: t / num_anios for m, t in ventas_historicas.items()}
max_prom = max(promedio_mensual.values())
mejores_meses_historicos = sorted(m for m, p in promedio_mensual.items() if p == max_prom)

print(f"""
Pregunta 1: Órdenes únicas: {q1}
Pregunta 2: Promedio de artículos por orden: {q2:.2f}
Pregunta 3: Máximo de artículos por orden: {q3}
Pregunta 4: Órdenes en octubre de 2021: {q4}
Pregunta 5: Cliente que más gastó en 2021: {', '.join(etiqueta(c) for c in clientes_top)} ({maximo_gasto:.2f})
Pregunta 6: Mejor mes histórico de ventas:
  - mes ganador más frecuente entre años: {', '.join(meses[m] for m in meses_mas_frecuentes)} ({max_frec} de {num_anios} años)
  - mayor promedio histórico de ventas por mes: {', '.join(meses[m] for m in mejores_meses_historicos)}
""")
```
